```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict
from langchain_google_genai import ChatGoogleGenerativeAI
from dotenv import load_dotenv
```

```python
load_dotenv()

model = ChatGoogleGenerativeAI(model = "gemini-2.5-flash", temperature= 0)
```

```python
### START -> outline -> blog content -> evaluation score -> END
class Blog(TypedDict):
    title: str
    outline: str
    content: str
    score: int
```

```python
def generate_outline(state: Blog) -> Blog:
    title = state['title']

    prompt = f"Generate a detailed outline for a blog on the topic: {title}"

    outline = model.invoke(prompt).content

    state['outline'] = outline

    return state
```

```python
def generate_blog(state: Blog) -> Blog:
    title = state['title']
    outline = state['outline']

    prompt = f"Generate a detailed blog on the topic: {title} using the following outline\n{outline}"

    blog = model.invoke(prompt).content

    state['content'] = blog

    return state
```

```python
def evaluate_blog(state: Blog) -> Blog:
    title = state['title']
    outline = state['outline']
    blog = state['content']

    prompt = f"topic: {title}\n{outline}\nBased on the above outline evaluate the following blog and give it an integer score out of 5\n{blog}"

    score = model.invoke(prompt).content

    state['score'] = score

    return state
```

```python
graph = StateGraph(Blog)

#nodes
graph.add_node("generate_outline", generate_outline)
graph.add_node("generate_blog", generate_blog)
graph.add_node("evaluate_blog", evaluate_blog)

# edges
graph.add_edge(START, "generate_outline")
graph.add_edge("generate_outline", "generate_blog")
graph.add_edge("generate_blog", "evaluate_blog")
graph.add_edge("evaluate_blog", END)

# compile
workflow = graph.compile()
```

```python
initial_state = {"title":"Effects of Tariffs on India"}

final_state = workflow.invoke(initial_state)
```
